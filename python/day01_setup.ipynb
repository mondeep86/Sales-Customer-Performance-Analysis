```python
query = """
WITH Order_Sales AS (
    SELECT
        "Order ID",
        SUM("Sales") AS Order_Total
    FROM superstore
    GROUP BY "Order ID"
)
SELECT
    COUNT(*) AS Total_Orders,
    ROUND(SUM(Order_Total), 2) AS Total_Revenue,
    ROUND(AVG(Order_Total), 2) AS Average_Order_Value
FROM Order_Sales;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
     "Discount",
     COUNT(*) AS Total_Records,
     ROUND(SUM("Sales"), 2) AS Total_Sales,
     ROUND(SUM("Profit"), 2) AS Total_Profit,
     ROUND(AVG("Profit"), 2) AS Average_Profit
FROM superstore
GROUP BY "Discount"
ORDER BY "Discount";
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Category_Performance AS (
    SELECT
        "Category",
        SUM("Sales") AS Total_Sales,
        SUM("Profit") as Total_Profit
    FROM superstore
    GROUP BY "Category"
)
SELECT
    "Category",
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(Total_Profit, 2) AS Total_Profit,
    ROUND(
        (Total_Profit / NULLIF(Total_Sales, 0)) * 100,
        2
    ) AS Profit_Margin_Percentage
FROM Category_Performance
ORDER BY Profit_Margin_Percentage DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Product_Sales AS (
    SELECT
        "Category",
        "Product Name",
        SUM("Sales") AS Total_Sales
    FROM superstore
    GROUP BY "Category", "Product Name"
    ),
    Ranked_Products AS (
        SELECT
            "Category",
            "Product Name",
            ROUND(Total_Sales, 2) AS Total_Sales,
            ROW_NUMBER() OVER(
                PARTITION BY "Category"
                ORDER BY Total_Sales DESC
            ) AS Product_Rank
        FROM Product_Sales
)
SELECT *
FROM Ranked_Products
WHERE Product_Rank <= 3
ORDER BY "Category", Product_Rank;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Category_Profit AS (
    SELECT
        "Category",
        SUM("Profit") AS Total_Profit
    FROM superstore
    GROUP BY "Category"
)
SELECT 
    "Category",
    ROUND(Total_Profit, 2) AS Total_Profit,
    RANK() OVER (
        ORDER BY Total_Profit DESC
    ) AS Profit_Rank, 
    DENSE_RANK() OVER (
        ORDER BY Total_Profit DESC
    ) AS Dense_Profit_Rank
FROM Category_Profit
ORDER BY Total_Profit DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
SELECT
    c."Segment",
    COUNT(DISTINCT c. "Customer ID") AS Total_Customers,
    COUNT(DISTINCT o."Order ID") AS Total_Orders,
    ROUND(SUM(s."Sales"), 2) AS Total_Sales,
    ROUND(SUM(s."Profit"), 2) AS Total_Profit
FROM customers c
JOIN orders o
    ON c."Customer ID" = o."Customer ID"
JOIN superstore s
    ON o."Order ID" = S."Order ID"
GROUP BY c."Segment"
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Customer_Orders AS (
    SELECT
        "Customer ID",
        COUNT(DISTINCT "Order ID") AS Total_Orders
    FROM orders
    GROUP BY "Customer ID"
),
Customer_Category AS (
    SELECT
        "Customer ID",
        Total_Orders,
        CASE
            WHEN Total_Orders >= 10 THEN 'High Frequency'
            WHEN Total_Orders >= 5 THEN 'Medium Frequency'
            ELSE 'Low Frequency'
        END AS Customer_Type
    FROM Customer_Orders
)
SELECT
    Customer_Type,
    COUNT(*) AS Number_of_Customers,
    ROUND(AVG(Total_Orders), 2) AS Average_Orders
FROM Customer_Category
GROUP BY Customer_Type
ORDER BY Number_of_Customers DESC;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Monthly_Sales AS (
    SELECT 
        SUBSTR("Order Date", 1, 7) AS Sales_Month,
        SUM("Sales") AS Total_Sales
    FROM superstore
    GROUP BY Sales_Month
),
Sales_Comparison AS (
    SELECT
        Sales_Month,
        Total_Sales,
        LAG(Total_Sales) OVER (
            ORDER BY Sales_Month
        ) AS Previous_Month_Sales
    FROM Monthly_Sales
)
SELECT 
    Sales_Month,
    ROUND(Total_Sales, 2) AS Total_Sales,
    ROUND(Previous_Month_Sales, 2) AS Previos_Month_Sales,
    ROUND(
        (Total_Sales - Previous_Month_Sales)
        / NULLIF(Previous_Month_Sales, 0) * 100,
        2
    ) AS Growth_Percentage
FROM Sales_Comparison
ORDER BY Sales_Month;
"""

result = pd.read_sql_query(query, conn)
display(result)
```

```python
query = """
WITH Monthly_Sales AS (
    SELECT 
        SUBSTR("Order Date", 1, 7) AS Sales_Month,
        SUM("Sales") AS Total_Sales
    FROM superstore
    GROUP BY Sales_Month
)
SELECT
    Sales_Month,
    ROUND(Total_Sales, 2) AS Monthly_Sales,
    ROUND(
        SUM(Total_Sales) OVER (
            ORDER BY Sales_Month
            ROWS BETWEEN UNBOUNDED PRECEDING
            AND CURRENT ROW
        ), 2
    ) AS Cumulative_Sales
FROM Monthly_Sales
ORDER BY Sales_Month;
"""

result = pd.read_sql_query(query, conn)
display(result)
```
